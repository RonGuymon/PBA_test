__Lab Insight:__ Are revenues dependent on advertising expenses? The more consumers know or are reminded about a product and its qualities via advertising, the more of that product they supposedly will buy. However, it isn’t always clear how big that impact is. Each company will be interested in knowing what the returns are to the advertising expenses for their specific firm. For this reason, companies will evaluate the relationship both by 1) using a visualization (scatter plot) and 2) running a regression to further understand that relationship. The outcome variable is sales revenue that arguably comes as a result of the input variable, advertising expenses.  
* Dependent variable: Sales revenue
* Independent variable: Advertising expenses

__Keywords:__ advertising, sales revenue

## Ask the Question
What is the relationship between advertising expense and sales for our company?

## Master the Data
The Lab 4.6 Data.xlsx file contains advertising expense and sales revenue over the past 36 months. The data dictionary for this file is below. 

|Column Name|Description| 
|:---|:---| 
|__Month__| The month number from the beginning of the analysis until the end of the analysis. |
|__Date__| Month when revenues and advertising expenses are realized|
|__Revenues__| The monthly revenue for this company| 
|__Advertising Expenses__| The monthly advertising expenses for this company|

## Perform the Analysis
### Analysis Task 1: Load the data

```python
import pandas as pd                      # Data manipulation
import seaborn as sns                    # Visualization
import statsmodels.formula.api as smf    # Regression analysis

# Load and explore the data
df = pd.read_excel('Lab 4.6 Data.xlsx')
display(df.info(), df.head())
```

### Analysis Task 2: Create a scatter plot evaluating the relationship between advertising expenditures and sales revenue

```python
sns.regplot(data = df, x='Advertising Expenses', y='Revenues', ci=False);
```

### Analysis Task 3: Using regression analysis, determine the relationship between advertising expenditures and sales revenue

```python
smf.ols("Revenues ~ Q('Advertising Expenses')", data = df).fit().summary()
```
