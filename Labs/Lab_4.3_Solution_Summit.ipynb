# Lab 4.3 - Exploring Data with Visualizations - Solution

## 1. Read the dataset into a pandas dataframe, filter it, and visually explore it


```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/RonGuymon/PBA_test/refs/heads/main/Labs/Lab_4.3_Data.csv')
df2 = df[df.world_rank <= 65][['company', 'icb3_short', 'region', 
                               'world_rank', 'rnd_22', 'netsales_22', 'emp']]
pd.options.display.precision = 2
display(df2.info(), df2)
```

## 2. Create a bar chart to explore the number of observations from each industry

```python
df2.icb3_short.value_counts().plot.barh();
```

## 3. Explore the distribution of R&D expenditures with box and whisker plots

```python
df2.rnd_22.plot.box();
```

```python
df2.plot.box(by='icb3_short', column='rnd_22', vert=False);
```

## 4. Explore the relationship between R&D expenditures and net sales with a scatter plot

```python
df2.plot.scatter(x='netsales_22', y='rnd_22');
```
